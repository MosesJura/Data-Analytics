```python
import random

products = ['Laptop', 'Monitor', 'Keyboard', 'Mouse', 'Webcam',
            'Headset', 'Docking Station', 'USB Hub', 'Desk Lamp', 'Surge Protector']

# 4a — Product of the Day
product_of_day = random.choice(products)
print(f"Product of the Day: {product_of_day}")

# 4b — 3 products for usability survey (no repeats)
survey_products = random.sample(products, 3)
print(f"Survey products: {survey_products}")

# 4c — shuffle for presentation (in place!)
random.shuffle(products)
print(f"Presentation order: {products}")
# NOTE: don't do x = random.shuffle(...) — x would be None

# 4d — simulated daily transaction count
transactions = random.randint(50, 300)
print(f"Simulated daily transactions: {transactions}")
```
