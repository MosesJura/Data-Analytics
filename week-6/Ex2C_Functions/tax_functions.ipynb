```python
# get_soc_sec_tax()
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062  
```

```python
# get_medicare_tax()
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
# Cell: get_federal_tax()
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:  # 4+
        rate = 0.18 - (withholding_code - 4) * 0.005
        # example: code 6 → 0.18 - 2*0.005 = 0.17
    return gross_pay * rate


    # Person1: gross 750, code 0
print(f"Person1 — SS: {get_soc_sec_tax(750):.2f}, "
      f"Medicare: {get_medicare_tax(750):.2f}, "
      f"Federal: {get_federal_tax(750, 0):.2f}")

# Person2: gross 1550, code 2
print(f"Person2 — SS: {get_soc_sec_tax(1550):.2f}, "
      f"Medicare: {get_medicare_tax(1550):.2f}, "
      f"Federal: {get_federal_tax(1550, 2):.2f}")

# Person3: gross 1100, code 5
print(f"Person3 — SS: {get_soc_sec_tax(1100):.2f}, "
      f"Medicare: {get_medicare_tax(1100):.2f}, "
      f"Federal: {get_federal_tax(1100, 5):.2f}")
```
