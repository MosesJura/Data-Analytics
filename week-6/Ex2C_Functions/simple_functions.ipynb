## Creating and calling functions

```python
# Cell: favorite_things()
def favorite_things(name, movie, music):
    return (f"Name: {name}\n"
            f"Favorite movie: {movie}\n"
            f"Favorite musician/band: {music}")

print(favorite_things("Alex", "Inception", "Radiohead"))


# Cell: why_im_here()
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined to build data skills for a career change."))


# Cell: favorite_place()
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."

# Test 1
print(favorite_place("Kyoto", "the temples and food are incredible"))

# Test 2
print(favorite_place("the Oregon coast", "I love the cold ocean and quiet beaches"))
```
