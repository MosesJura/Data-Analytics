```python
# Cell: get_supplier_code()
def get_supplier_code(part_code):
    colon = part_code.find(':')
    return part_code[:colon]

# Cell: get_product_num()
def get_product_num(part_code):
    colon = part_code.find(':')
    dash = part_code.find('-')
    return part_code[colon+1:dash]

# Cell: get_size()
def get_size(part_code):
    dash = part_code.find('-')
    return part_code[dash+1:]
```

```python
code_1 = 'ACME:123-L'
code_2 = 'DI:12345-M'
code_3 = 'ACE:1-12'
```

```python
print(f"For {code_1}, the supplier is {get_supplier_code(code_1)}. "
      f"Part number {get_product_num(code_1)} is available in size {get_size(code_1)}.")
# For ACME:123-L, the supplier is ACME. Part number 123 is available in size L.
```
