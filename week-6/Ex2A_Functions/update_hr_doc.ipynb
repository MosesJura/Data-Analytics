## Lab 3. Employee data
This lab exercise makes updates to information extracted from an employee database.

```python
hr_list = [
    ('0123', 'HR',  'Rebecca Yang',  '69000'),
    ('0121', 'IT',  'Mark Blick',    '67000'),
    ('0068', 'IT',  'Kahna Larsen',  '112000'),
    ('0234', 'OPS', 'Jim Smith',     '54000')
]
```

```python
# Tuple replacement: rebuild the whole tuple at index 1
hr_list[1] = ('0121', 'IT', 'Mark Blick-Hawley', '67000')
```

```python
print(hr_list)
```

```python
hr_list[3] = ('0234', 'CS', 'Jim Smith', '60000')
```

```python
print(hr_list)
```

```python
print("Employee# | DeptCode | Name | NN,NNN")
for emp in hr_list:
    emp_num, dept, name, salary = emp          # unpack the tuple
    salary_int = int(salary)                   # string → int
    print(f"{emp_num} | {dept} | {name} | {salary_int:,}")
```
