## Lab 1. Working with names - parsing strings


```python
test_name = 'John Doe'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")

# Note: running this cell produces NO output —
# the function is defined but not called. That's expected.
```

```python
parse_and_display(test_name)
# Full name: John Doe
# First name: John
# Last name: Doe
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[0])
parse_and_display(test_list[1])
parse_and_display(test_list[2])
```

```python
parse_and_display('Moses Jura')
```

```python
def parse_and_display2(name):
    parts = name.split()          # list of words

    if len(parts) == 1:
        print(f"Full name: {name}")
        print(f"First name: {parts[0]}")
        print("Last name: (none)")
    elif len(parts) == 2:
        print(f"Full name: {name}")
        print(f"First name: {parts[0]}")
        print(f"Last name: {parts[1]}")
    else:  # 3+ words
        print(f"Full name: {name}")
        print(f"First name: {parts[0]}")
        print(f"Middle name: {' '.join(parts[1:-1])}")
        print(f"Last name: {parts[-1]}")

# Tests:
parse_and_display2('Lorde')                  # 1 word
parse_and_display2('Billie Eilish')          # 2 words
parse_and_display2('Megan Thee Stallion')    # 3 words
```
